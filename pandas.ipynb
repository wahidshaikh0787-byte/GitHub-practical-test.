```python
import pandas as pd
import numpy as np
```

isnull() and notnull()

```python
df = pd.DataFrame({
    "Name": ["Aman", "Riya", None, "Karan"],
    "Age": [20, np.nan, 22, 21],
    "Marks": [85, 90, np.nan, 78]
})

print(df)

# Check every cell for missing values
print(df.isnull())

# Check every cell for non-missing values
print(df.notnull())

# Count missing values column-wise
print(df.isnull().sum())

# Count total missing values
print(df.isnull().sum().sum())

# Display rows where Age is missing
print(df[df["Age"].isnull()])

# Display rows where Age is available
print(df[df["Age"].notnull()])
```

Dropping missing values: dropna()

```python
df = pd.DataFrame({
    "Name": ["Aman", "Riya", "Karan", "Neha"],
    "Age": [20, np.nan, 22, 21],
    "Marks": [85, 90, np.nan, 78]
})

# Remove rows containing at least one null value
print(df.dropna())

# Remove columns containing at least one null value
print(df.dropna(axis=1))

# Remove rows only when all values are null
print(df.dropna(how="all"))

# Remove rows if Age is missing
print(df.dropna(subset=["Age"]))

# Remove rows having fewer than 2 non-null values
print(df.dropna(thresh=2))
```

Imputing missing values: fillna() and strategies

```python
df = pd.DataFrame({
    "Name": ["Aman", "Riya", "Karan", "Neha"],
    "Age": [20, np.nan, 22, 21],
    "Marks": [85, 90, np.nan, 78]
})

# Replace all missing values with 0
print(df.fillna(0))

# Replace missing values in one column
df["Age"] = df["Age"].fillna(21)

# Mean imputation
df["Marks"] = df["Marks"].fillna(df["Marks"].mean())

print(df)
```

Other common strategies:

```python
df = pd.DataFrame({
    "Name": ["Aman", "Riya", "Karan", "Neha"],
    "Age": [20, np.nan, 22, 21],
    "Marks": [85, 90, np.nan, 78],
    "Department": ["CSE", "IT", None, "CSE"]
})

# Mean: useful for reasonably symmetric numeric data
df["Age"] = df["Age"].fillna(df["Age"].mean())

# Median: often better when numeric data has outliers
df["Age"] = df["Age"].fillna(df["Age"].median())

# Mode: useful for categorical data
df["Department"] = df["Department"].fillna(
    df["Department"].mode()[0]
)

# Forward fill: use the previous available value
df["Age"] = df["Age"].ffill()

# Backward fill: use the next available value
df["Age"] = df["Age"].bfill()

print(df)
```

map() — transform values in a column

```python
df = pd.DataFrame({
    "Name": ["Aman", "Riya", "Karan"],
    "Grade": ["A", "B", "A"]
})

# Convert grades into grade points
grade_points = {
    "A": 10,
    "B": 8,
    "C": 6
}

df["Grade_Points"] = df["Grade"].map(grade_points)

print(df)
```

apply() — perform calculations on columns or rows

```python
df = pd.DataFrame({
    "Name": ["Aman", "Riya", "Karan"],
    "Maths": [80, 90, 70],
    "Python": [90, 85, 75]
})

# Apply a function to each value in a column
df["Maths_Extra"] = df["Maths"].apply(lambda x: x + 5)

# Calculate average marks for each student
df["Average"] = df[["Maths", "Python"]].apply(
    lambda row: row.mean(),
    axis=1
)

# Apply a function to each column
print(df[["Maths", "Python"]].apply(np.mean, axis=0))

print(df)
```

DataFrame map() — transform every cell

```python
df = pd.DataFrame({
    "Maths": [80, 90, 70],
    "Python": [60, 85, 95]
})

# Add 5 to every numeric cell
result = df.map(lambda x: x + 5)

print(result)
```

sort_values()

```python
df = pd.DataFrame({
    "Name": ["Aman", "Riya", "Karan", "Neha"],
    "Marks": [85, 95, 75, 95],
    "Age": [20, 21, 19, 20]
})

# Sort marks in ascending order
print(df.sort_values(by="Marks"))

# Sort marks in descending order
print(df.sort_values(by="Marks", ascending=False))

# Sort by marks, then by age
print(df.sort_values(by=["Marks", "Age"]))

# Sort in descending order using multiple columns
print(df.sort_values(
    by=["Marks", "Age"],
    ascending=[False, True]
))
```

String operations using .str

```python
df = pd.DataFrame({
    "Name": [" aman rai ", "RIYA SHARMA", "karan singh"],
    "Email": [
        "aman@gmail.com",
        "riya@yahoo.com",
        "karan@gmail.com"
    ]
})

# Convert text to uppercase
df["Upper_Name"] = df["Name"].str.upper()

# Convert text to lowercase
df["Lower_Name"] = df["Name"].str.lower()

# Remove spaces from beginning and end
df["Clean_Name"] = df["Name"].str.strip()

# Capitalize names
df["Title_Name"] = df["Clean_Name"].str.title()

# Count characters
df["Name_Length"] = df["Clean_Name"].str.len()

# Check whether email contains gmail.com
df["Is_Gmail"] = df["Email"].str.contains(
    "gmail.com",
    regex=False,
    na=False
)

# Extract email provider
df["Provider"] = df["Email"].str.split("@").str[-1]

# Replace spaces with underscores
df["Name_ID"] = df["Clean_Name"].str.replace(
    " ", "_", regex=False
)

print(df)
```

groupby() and split–apply–combine

```python
df = pd.DataFrame({
    "Name": ["Aman", "Riya", "Karan", "Neha", "Vikas"],
    "Department": ["CSE", "IT", "CSE", "IT", "CSE"],
    "Marks": [85, 90, 75, 95, 80]
})

# Average marks for each department
print(df.groupby("Department")["Marks"].mean())

# Total marks for each department
print(df.groupby("Department")["Marks"].sum())

# Number of students in each department
print(df.groupby("Department")["Name"].count())

# Average and maximum marks
print(df.groupby("Department")["Marks"].agg(["mean", "max"]))

# Group by department and calculate average marks and count
print(
    df.groupby("Department")
      .agg(
          Average_Marks=("Marks", "mean"),
          Maximum_Marks=("Marks", "max"),
          Student_Count=("Name", "count")
      )
)

# Group using multiple columns
df["Year"] = [1, 1, 2, 2, 1]

print(
    df.groupby(["Department", "Year"])["Marks"].mean()
)
```

agg() with multiple functions

```python
df = pd.DataFrame({
    "Department": ["CSE", "CSE", "IT", "IT", "CSE"],
    "Marks": [85, 90, 75, 95, 80],
    "Attendance": [90, 95, 80, 85, 88]
})

# Multiple functions on one column
print(
    df.groupby("Department")["Marks"].agg(
        ["min", "max", "mean", "sum", "count"]
    )
)

# Different functions for different columns
summary = df.groupby("Department").agg(
    Average_Marks=("Marks", "mean"),
    Highest_Marks=("Marks", "max"),
    Average_Attendance=("Attendance", "mean"),
    Student_Count=("Marks", "count")
)

print(summary)
```

pivot_table() and crosstab()

```python
df = pd.DataFrame({
    "Department": ["CSE", "CSE", "IT", "IT", "CSE", "IT"],
    "Year": [1, 2, 1, 2, 1, 2],
    "Marks": [80, 90, 75, 85, 95, 90]
})

# Average marks by department and year
table = pd.pivot_table(
    df,
    values="Marks",
    index="Department",
    columns="Year",
    aggfunc="mean"
)

print(table)
```

crosstab() — frequency counts

```python
df = pd.DataFrame({
    "Department": ["CSE", "CSE", "IT", "IT", "CSE", "IT"],
    "Gender": ["F", "M", "F", "M", "M", "F"],
    "Result": ["Pass", "Pass", "Fail", "Pass", "Fail", "Pass"]
})

# Count students by department and gender
print(pd.crosstab(df["Department"], df["Gender"]))

# Count department-wise pass/fail results
print(pd.crosstab(df["Department"], df["Result"]))

# Show proportions instead of counts
print(pd.crosstab(
    df["Department"],
    df["Result"],
    normalize="index"
))
```

Finding and removing duplicates

```python
df = pd.DataFrame({
    "Student_ID": [101, 102, 101, 103, 102],
    "Name": ["Aman", "Riya", "Aman", "Karan", "Riya"],
    "Marks": [80, 90, 80, 75, 90]
})

# Identify duplicate rows
print(df.duplicated())

# Count duplicate rows
print(df.duplicated().sum())

# Display duplicate rows
print(df[df.duplicated()])

# Remove duplicate rows
print(df.drop_duplicates())

# Remove duplicates based only on Student_ID
print(df.drop_duplicates(subset=["Student_ID"]))

# Keep the last occurrence instead of the first
print(df.drop_duplicates(
    subset=["Student_ID"],
    keep="last"
))

# Update the original DataFrame
df = df.drop_duplicates()
print(df)
```

Extracting year, month, day and weekday

```python
df = pd.DataFrame({
    "Student": ["Aman", "Riya", "Karan"],
    "Admission_Date": [
        "2025-07-15",
        "2025-08-20",
        "2026-01-10"
    ]
})

df["Admission_Date"] = pd.to_datetime(df["Admission_Date"])

# Extract year
df["Year"] = df["Admission_Date"].dt.year

# Extract month number
df["Month"] = df["Admission_Date"].dt.month

# Extract day of month
df["Day"] = df["Admission_Date"].dt.day

# Extract weekday name
df["Weekday"] = df["Admission_Date"].dt.day_name()

# Extract weekday number: Monday=0, Sunday=6
df["Weekday_Number"] = df["Admission_Date"].dt.dayofweek

print(df)
```

Filter records within a date range

```python
df = pd.DataFrame({
    "Date": [
        "2025-01-05",
        "2025-01-15",
        "2025-02-10",
        "2025-03-20",
        "2025-04-05"
    ],
    "Sales": [1000, 1500, 2000, 1800, 2500]
})

df["Date"] = pd.to_datetime(df["Date"])

# Select January records
january = df[
    (df["Date"] >= "2025-01-01") &
    (df["Date"] < "2025-02-01")
]

print(january)

# Select records between two dates, including both dates
filtered = df[df["Date"].between("2025-02-01", "2025-03-31")]

print(filtered)
```

resample() — aggregate data over time intervals

```python
df = pd.DataFrame({
    "Date": pd.date_range(
        start="2025-01-01",
        periods=90,
        freq="D"
    ),
    "Sales": np.random.randint(100, 1000, size=90)
})

# Set the date as the index
df = df.set_index("Date")

# Monthly total sales
monthly_sales = df["Sales"].resample("MS").sum()
print(monthly_sales)

# Weekly average sales
weekly_sales = df["Sales"].resample("W").mean()
print(weekly_sales)

# Quarterly total sales
quarterly_sales = df["Sales"].resample("QS").sum()
print(quarterly_sales)

# Monthly minimum and maximum sales
monthly_summary = df["Sales"].resample("MS").agg(["min", "max", "mean", "sum"])
print(monthly_summary)
```
